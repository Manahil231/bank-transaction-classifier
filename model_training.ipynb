## Bank Transaction – End-to-End Machine Learning Project

**Goal:** Take raw daily bank files, clean them, explore them, and then train a Machine Learning model that can look at a transaction and **predict whether it is a Deposit or a Withdrawal**. At the end we save the whole workflow as `bank_transaction.pkl` and prove that the saved file works.

**Why this target?** `Transaction_Type` (Deposit / Withdrawal) is a clear two-option answer, and the other columns (amount, branch, account, date) are the "clues" the model uses. This is called a **binary classification** problem.

## 2. Load and Understand the Data

Humare paas **do tarah ki files** hain:

| File | Kya hai |
|---|---|
| `data/daily/*.csv` | Har din ki alag file (91 files): Date, Account_ID, Transaction_Type, Amount, Branch |
| `data/accounts.csv` | Customers ki list: Account_ID, Customer_Name, Account_Type, Branch |

**Simple words:** Daily files bank ki rozana entries hain. `accounts.csv` customers ki list hai. Dono ko `Account_ID` se jorte hain.

**Step 2.1:** Pehle saari daily files ko ek table (DataFrame) mein jama karte hain.

Daily files mein koi ID column nahi hai, is liye hum khud har transaction ko **unique `Transaction_ID`** dete hain (jaise `2026-01-01-001` = 1 January ki pehli entry). Is se har row ki alag pehchan ho jati hai.

```python
# data/daily folder ki saari .csv files ki list banao
daily_files = sorted(Path("data/daily").glob("*.csv"))
print("Total daily files:", len(daily_files))

# Har file ko padho aur ek list mein rakho
all_days = []
for file in daily_files:
    one_day = pd.read_csv(file)
    one_day["Source_File"] = file.stem      # file ka naam (jaise 2026-01-01) yaad rakhne ke liye
    all_days.append(one_day)

# Saari chhoti tables ko ek badi table mein jodo
transactions = pd.concat(all_days, ignore_index=True)

# Har din ke andar 1, 2, 3... number do, phir date ke saath jod kar unique ID banao
transactions["Row_No"] = transactions.groupby("Source_File").cumcount() + 1
transactions["Transaction_ID"] = transactions["Source_File"] + "-" + transactions["Row_No"].astype(str).str.zfill(3)

# Customers ki file padho
accounts = pd.read_csv("data/accounts.csv")

print("Transactions shape:", transactions.shape)
print("Accounts shape    :", accounts.shape)
```

**Step 2.3:** `accounts` ko `transactions` ke saath jorte hain (merge). Hum sirf `Account_Type` lete hain, kyunke:
- `Customer_Name` har account ka alag naam hai (ye sirf pehchan hai, model ko kuch sikhata nahi).

#### 6) Outliers (IQR method)
Rule: jo value `Q1 - 1.5*IQR` se kam ya `Q3 + 1.5*IQR` se zyada ho, wo outlier hai.

```python
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Amount"] < lower) | (df["Amount"] > upper)]
print("Lower limit:", lower, "| Upper limit:", upper)
print("Number of outliers:", len(outliers))
print("Invalid amounts (zero or negative):", (df["Amount"] <= 0).sum())
```

## 3. Data Cleaning

Findings ke hisaab se cleaning ke faislay:

1. **Duplicate check ID par karenge** → Same-content wali dono rows ki `Transaction_ID` alag hai, yani ho sakta hai ke ek hi din ek account se ek hi amount do baar nikala gaya ho (asli transactions). Pakka nahi keh sakte ke ghalti hai, is liye **dono rows rakhte hain**. 878 mein se sirf 1 row ka farq hai, model par asar na ke barabar hai. Sirf wohi row hatayenge jiski `Transaction_ID` repeat ho.
2. **`Date` ko datetime banayen** → taake month, day, weekday nikal saken.
3. **Missing values** → hain hi nahi, is liye kuch bharna nahi. (Phir bhi pipeline mein imputer lagayenge taake future mein koi missing value aaye to model crash na kare.)
4. **Outliers** → 0 hain, is liye kuch nahi hataya. Amount ki skewness bhi kam hai, to log transform ki zaroorat nahi.
5. **Naye features** → `Month`, `Day`, `DayOfWeek` (Date se).

```python
df_clean = df.copy()

# 1) Sirf wo rows hatao jin ki Transaction_ID repeat ho (abhi koi nahi)
before = len(df_clean)
df_clean = df_clean.drop_duplicates(subset="Transaction_ID").reset_index(drop=True)
print("Rows removed:", before - len(df_clean))

# 2) Date ko datetime banao
df_clean["Date"] = pd.to_datetime(df_clean["Date"])

# 3) Date se naye columns banao
df_clean["Month"] = df_clean["Date"].dt.month
df_clean["Day"] = df_clean["Date"].dt.day
df_clean["DayOfWeek"] = df_clean["Date"].dt.dayofweek     # 0 = Monday ... 6 = Sunday

print("Clean shape:", df_clean.shape)
df_clean.head()
```

## Prepare Features and Target

- **Target (y):** `Transaction_Type` — jo model ko predict karna hai.
- **Features (X):** baqi useful columns.

**Kya nahi lenge?**
- `Date` (hum Month/Day/DayOfWeek le chuke hain), `Source_File` (sirf file ka naam, Date jaisa hi).
- `Account_ID` ko rakh sakte hain, magar yaad rahe ye sirf 10 accounts hain, is liye ise categorical column maanenge.

```python
target = "Transaction_Type"

numeric_cols = ["Amount", "Month", "Day", "DayOfWeek"]
categorical_cols = ["Account_ID", "Branch", "Account_Type"]

X = df_clean[numeric_cols + categorical_cols]
y = (df_clean[target] == "Withdrawal").astype(int)     # Deposit = 0, Withdrawal = 1

print("Numeric columns    :", numeric_cols)
print("Categorical columns:", categorical_cols)
print("X shape:", X.shape, "| y shape:", y.shape)
```

> **Dhyan dein:** Model sirf numbers samajhta hai, is liye target ko `Deposit = 0`, `Withdrawal = 1` bana diya.

## 4. Train-Test Split

Data ko 2 hisson mein baantte hain:
- **Train (80%)** → model isse seekhta hai.
- **Test (20%)** → model ka imtihan, jo usne pehle kabhi nahi dekha.

`random_state=42` rakhne se har baar wahi split milta hai. `stratify=y` se dono hisson mein Deposit/Withdrawal ka ratio barabar rehta hai.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

print("Train:", X_train.shape, "| Test:", X_test.shape)
print("Train target ratio:", y_train.mean().round(3), "| Test target ratio:", y_test.mean().round(3))
```

## Preprocessing

Raw data model ko seedha nahi de sakte. Do alag tareeqe:

| Column type | Kya karte hain | Kyun |
|---|---|---|
| **Numeric** | Missing ko median se bharo + `StandardScaler` | Sab numbers ek jaisi scale par aa jayen |
| **Categorical** | Missing ko sab se common value se bharo + `OneHotEncoder` | Text (Karachi, Lahore...) ko 0/1 columns mein badalna |

`ColumnTransformer` dono kaam ek saath sambhalta hai. `handle_unknown="ignore"` ka matlab: agar naya branch aaye to model crash nahi hoga.

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_cols),
    ("cat", categorical_pipeline, categorical_cols),
])

preprocessor
```

---
## Train Models and Compare

Hum 6 models try karte hain:

1. **Dummy** — sirf sab se common class bata deta hai (ye *baseline* hai; koi bhi model isse behtar hona chahiye).
2. **Logistic Regression** — simple aur tez.
3. **KNN** — qareeb ke mushabeh data dekh kar faisla.
4. **Decision Tree** — sawalon ka darakht (if-else).
5. **Random Forest** — bohat se trees ka group.
6. **Gradient Boosting** — trees jo ek doosre ki ghaltiyan theek karte hain.

Har model ko `Pipeline` mein `preprocessor` ke saath rakha hai, taake preprocessing + model **ek hi object** ban jaye (yehi baad mein save hoga).

Hum do tarah se naapte hain:
- **CV ROC-AUC:** Train data ko 5 hisson mein baant kar average (zyada bharosemand).
- **Test score:** Alag rakhe hue test data par.

```python
models = {
    "Dummy (Baseline)": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=RANDOM_STATE),
    "KNN": KNeighborsClassifier(n_neighbors=7),
    "Decision Tree": DecisionTreeClassifier(max_depth=4, random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=6, random_state=RANDOM_STATE),
    "Gradient Boosting": GradientBoostingClassifier(random_state=RANDOM_STATE),
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
results = []
trained_pipelines = {}

for name, model in models.items():
    pipe = Pipeline([("preprocess", clone(preprocessor)), ("model", model)])

    # 5-fold cross validation (sirf train data par)
    cv_auc = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="roc_auc").mean()

    # Train par seekho, test par check karo
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    auc = roc_auc_score(y_test, pipe.predict_proba(X_test)[:, 1])

    trained_pipelines[name] = pipe
    results.append({
        "Model": name,
        "CV ROC-AUC": round(cv_auc, 3),
        "Accuracy": round(accuracy_score(y_test, y_pred), 3),
        "Precision": round(precision_score(y_test, y_pred, zero_division=0), 3),
        "Recall": round(recall_score(y_test, y_pred, zero_division=0), 3),
        "F1": round(f1_score(y_test, y_pred, zero_division=0), 3),
        "ROC-AUC": round(auc, 3),
    })
    print("Done:", name)
```

## Model Evaluation

**Metrics ka matlab (asaan alfaaz mein):** (Yahan "positive" class = Withdrawal)
- **Accuracy:** Kitne predictions sahi nikle.
- **Precision:** Jab model ne "Withdrawal" kaha, to kitni baar sahi tha.
- **Recall:** Asli Withdrawals mein se kitne model ne pakre.
- **F1:** Precision aur Recall ka balance.
- **ROC-AUC:** Model Deposit/Withdrawal ko alag karne mein kitna achha hai (0.5 = andaza, 1.0 = perfect).

## Select the Final Model

**Selection ka rule:** Hum **CV ROC-AUC** (train data ke 5-fold average) se model chunte hain, test score se nahi. Test data ko sirf aakhri imtihan ke liye rakhte hain, warna hum test ko dekh kar model chun lenge aur score jhootha ho jayega.

```python
best_name = results_df.loc[0, "Model"]
print("Final model:", best_name)

final_model = trained_pipelines[best_name]
y_pred = final_model.predict(X_test)

print()
print(classification_report(y_test, y_pred, target_names=["Deposit", "Withdrawal"]))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=["Deposit", "Withdrawal"], cmap="Blues")
plt.title("Confusion Matrix - " + best_name)
plt.show()
```

**Humne ye model kyun chuna?**
1. Is ka **CV ROC-AUC sab se zyada** hai (ROC-AUC kisi ek threshold par depend nahi karta, is liye model chunne ke liye mazboot metric hai), yani alag alag data hisson par bhi achha chala (sirf ek lucky split par nahi).
2. Ye Dummy baseline se behtar hai.
3. Table mein CV ROC-AUC aur test ROC-AUC ko compare karein: agar dono qareeb hon to model **overfit nahi** hua.

**Confusion matrix kaise padhein:** Diagonal (upar-left aur neeche-right) sahi predictions hain, baqi ghaltiyan.

> **Model ki kamzori (imaandari se):** Confusion matrix dekhein. Decision Tree ka **Recall bohat zyada** hai (lagbhag har Withdrawal pakar leta hai) lekin **Precision kam** hai. Matlab ye zyada tar transactions ko "Withdrawal" keh deta hai, is liye bohat se Deposits ko galat Withdrawal bata deta hai. Gradient Boosting jaise models zyada balanced hain, magar un ka ROC-AUC Decision Tree ke bohat qareeb hai, yani in mein farq chhota hai. Humne selection ka rule pehle se tay kiya tha (CV ROC-AUC), is liye usi par qayam hain.

> **Imaandari ki baat:** Is dataset mein sirf Amount ka mazboot asar hai, is liye accuracy 100% nahi hogi. Ye normal hai. Real bank data mein description, balance jaise aur columns hote hain jo accuracy barhate hain.
